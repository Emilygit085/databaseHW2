```python
select orders.order_id, users.username, products.product_name, order_items.quantity, order_items.unit_price
from orders, users, order_items, products
where orders.user_id = users.user_id
 and orders.order_id = order_items.order_id
 and order_items.product_id = products.product_id
 and orders.status = '已完成';
```

```python
SELECT 
    o.order_id,
    u.username,
    p.product_name,
    oi.quantity,
    oi.unit_price
FROM orders o
JOIN users u ON o.user_id = u.user_id
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
WHERE o.status = '已完成';
```

# 2

```python
SELECT 
    u.user_id,
    u.username,
    SUM(o.total_amount) AS total_spent
FROM users u
JOIN orders o ON u.user_id = o.user_id
GROUP BY u.user_id, u.username
HAVING SUM(o.total_amount) > 1000
ORDER BY total_spent DESC;
```

# 3

```python
SELECT DISTINCT username, email
FROM users
WHERE user_id IN (
    SELECT DISTINCT o.user_id
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products p ON oi.product_id = p.product_id
    WHERE p.category = '数码'
);
```

```python
SELECT DISTINCT username, email
FROM users u
WHERE EXISTS (
    SELECT 1
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products p ON oi.product_id = p.product_id
    WHERE o.user_id = u.user_id
      AND p.category = '数码'
);
```

```python
-- 步骤1：创建视图
CREATE VIEW sales_nov_2024 AS
SELECT 
    p.category,
    p.product_name,
    oi.quantity,
    oi.unit_price
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id
JOIN products p ON oi.product_id = p.product_id
WHERE o.order_date >= '2024-11-01' 
  AND o.order_date < '2024-12-01';

-- 步骤2：基于视图查询
SELECT 
    category,
    SUM(quantity) AS total_sold
FROM sales_nov_2024
GROUP BY category
ORDER BY total_sold DESC
LIMIT 3;
```
